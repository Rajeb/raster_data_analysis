```python
import rasterio 
import numpy as np
import matplotlib.pyplot as plt
```

```python
# visualizing single band raster
raster_file = 'single_band_raster.tif'
with rasterio.open(raster_file) as dataset:
    band_data = dataset.read(1)
type(band_data)
```

```python
plt.imshow(band_data, cmap="Purples")
plt.colorbar(label= "Pixel Value")
```

```python
# visualizing RGB raster
raster_file = 'rgb_raster.tif'
with rasterio.open(raster_file) as dataset:
    red_data = dataset.read(1)
    blue_data = dataset.read(2)
    green_data = dataset.read(3)

    rgb_data = np.stack([red_data, green_data, blue_data], axis=-1)
```

```python
#
plt.imshow(rgb_data)
plt.colorbar(label= "Pixel Value")
```

```python
# each band separately
f, ax = plt.subplots(1,3, figsize=(15,4))
ax[0].imshow(red_data, cmap="Reds")
ax[0].axis('off')
ax[1].imshow(green_data, cmap="Greens")
ax[1].axis('off')
ax[2].imshow(blue_data, cmap="Blues")
ax[2].axis('off')
```
